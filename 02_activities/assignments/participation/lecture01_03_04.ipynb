```python
#setting up a seed and data set
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
#creating a plot
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show() #in jupyter notebooks, fig.show() is not necessary
```

```python
#add a legend
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='upper center') #location of legend
fig.show()
```

```python
#modifying legend 
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='lower right',
          frameon = True, #add frame around the legend
          fontsize = 12, #change font size
          ncol = 2, #specify number of columns
          shadow = True )
fig.show()
```

```python
#relocating the legend to outside of the plot area :)
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='upper left', bbox_to_anchor=(1,1)) #will put it on the right top of the figure, this is sort of like a grid outside of a grid 
#0,0 would give bottom left outside of plot area
```

```python
#annotating with text() function
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person 2")
ax.legend(loc='lower right')

ax.text(10,95,"This value is important !", ha='center', color='red',size=20) #adjusting the text and adding text annotation
```

```python
#positioning our text annotations
fig, ax=plt.subplots()
ax.axis=([0,10,0,10]) #making set of empty axes
ax.text(1,5,". Data:(1,5)", transform=ax.transData) #referencing data coordinates
ax.text(0.5,0.1, ". Axes:(0.5,0.1)", transform=ax.transAxes) #referencing axes coordinates
ax.text(0.5,0.1,".Figure:(0.5,0.1)",transform=fig.transFigure) #referencing figure coordinates

#essentially, there are 3 coordinate systems: data, axes, figure
#data: coordinates based on the data points
#axes: coordinates based on the axes (0,0 is bottom left of axes, 1,1 is top right of axes)
#figure: coordinates based on the entire figure (0,0 is bottom left of figure, 1,1 is top right of figure)
```

```python
#annotate with arrows using the annotate() function
fig, ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person2")
ax.legend(loc='lower right') 
ax.annotate('This is important',xy=(10,95),xytext=(20,94),arrowprops=dict(facecolor='black')) #note that arrowprops is a dictionary, it takes key-value pairs
```

```python
#lets try some cool arrows 
fig, ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person2")
ax.legend(loc='lower right') 
ax.annotate('This is important',xy=(10,95),xytext=(20,94),arrowprops=dict(arrowstyle = "wedge",color="hotpink"))
```

```python
#removing ticks and labels
fig, ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person2")
ax.legend(loc='lower right') 

ax.yaxis.set_major_locator(plt.NullLocator()) #removes major ticks
ax.xaxis.set_major_formatter(plt.NullFormatter()) #removes major tick labels
```

```python
#limit tick marks :)
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person2")
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3)) #limits number of x-axis ticks to 3
```

```python
#tick mark intervals
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person2")
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(10)) #sets x-axis ticks at intervals of 10
```

```python
#rotate axis labels! 
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label="Person 1")
ax.scatter(x,y2,label="Person2")
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right') #modifying x-axis tick labels to be rotated 45 degrees, and right aligned
```

```python
#subplots, first import
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
#generate some random data (new data)
np.random.seed(613)
x1=np.arange(50)
y1=np.random.randint(0,75,50) 
x2=np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2=np.array([110,180,240,99,220])
```

```python
#creating subplot
fig, (ax1,ax2) = plt.subplots(ncols=2,nrows=1, figsize=(7,3)) #you essentially have 2 axes now, tuples
#now specify what goes in each subplot
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()

#alternative option, storing axes
#fig,ax = plt.subplots(ncols=2,nrows=1, figsize=(7,3))
#ax[0].scatter(x1,y1)
#ax[1].bar(x2,y2)
#fig.show()
```

```python
#plt.suplot_mosiac(), essentially a more flexible way of creating subplots
#you can name the axes however you want!
fig,someaxes=plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']], figsize=(7,4))
#referring to the specific subplots
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
plt.show() 
```

```python
#formatting our layout!
fig,someaxes=plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']], figsize=(7,4), layout='constrained') #these axis are organized in rows! 
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
someaxes['ax1'].set_xlabel('A Big Label',fontsize=18)
someaxes['ax2'].set_xlabel('Another Label',fontsize=18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18)
#fig.tight_layout() #tight layout to prevent overlapping

#differnce between the contrained layout and tight layout is that constrained layout automatically adjusts the subplot parameters to give specified padding, while tight layout only adjusts subplot params to give specified padding. Constrained layout is generally better for complex layouts. You can also include legends etc. 
```

```python
#multiple visualizations in one axes (sub-subplots)
#data first!
x=np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1=np.array([110,180,240,99,220])
y2=np.array([170,100,90,120,50])

fig,ax=plt.subplots(figsize=(7,3)) #just specify one axes

ax.bar(x,y1,color='indigo')
ax.plot(x,y2,color='red')
```

```python
#adding some error info
#calculate standard deeviation
y2_sd=np.std(y2)
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='blue') #this is just our plot! 
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='none') #fmt is for style/outlining dataa while yerr is for errorbars
```

```python
y2_sd=np.std(y2)
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='blue') #this is just our plot! 

#adding more options! 
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='none', ecolor='indigo',elinewidth=2, capsize=5,
            capthick=2, errorevery =2)

#errorevery is for skipping points! intervallic in many ways :)
```

```python
#Seaborn, data viz package built on top of matplotlib
#it is like matplotlib but helps us semantically map, where mapping means linking data variables to visual properties
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
import PIL
import requests
```

```python
#loading some data
tips = sns.load_dataset("tips")
print(tips)
#the tips dataset is from the sns
```

```python
#making a basic plot of tip amount v. total bill
sns.lineplot(data=tips,x='total_bill',y='tip')
```

```python
#using premade styles
sns.set_style('whitegrid')
sns.lineplot(data=tips,x='total_bill',y='tip')
```

```python
#add title and axis labels
tipgraph = sns.lineplot(data=tips,x='total_bill',y='tip')
tipgraph.set(title='Tips vs. Total Bill', xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
```

```python
#modifying aesthetic elements 
sns.set_style('ticks')
tipgraph=sns.lineplot(data=tips,
                      x='total_bill',
                      y='tip',
                      color='hotpink',
                      linestyle='--',
                      linewidth=3,
                      marker='o',
                      markerfacecolor='indigo')

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
#using multiple variables
tipgraph = sns.scatterplot(data=tips,x='total_bill',
                           y='tip',style='time',hue='day',
                           palette=['purple','hotpink','deepskyblue','yellowgreen'])


tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
#pairplot, a way to compare variables
sns.pairplot(data=tips) #pairplot creates a matrix of scatterplots for each pair of variables
#useful for exploratory data analysis 
```

```python
#pairplot, a way to compare variables
sns.pairplot(data=tips, hue='day') #segmented by day of the week
```

```python
#relplot - similar, explore different levels within variables 
daysplot=sns.relplot(data=tips,x='total_bill',
                    y='tip',
                    hue='sex', #seperate by sex in data points 
                    col='day', #column seperation by day 
                    kind='scatter', #type of plot
                    col_wrap=2) #elements per row 

#you can see relplot, as getting more specific with pairplot. You are looking into relationships b/w variables
```

```python
#plotly
#plotly is another data viz package, but it is for interactive visualizations
import plotly.graph_objects as go 
#some new data
x1 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110,180,240,99,220])
```

```python
#plotly's basic plot structure is different from matplotlib and seaborn
#make figure
graph = go.Figure()

#choose plot type and data
graph.add_trace(go.Bar(x=x1,y=y1))

#add titles
graph.update_layout(title='Pirate Scores',
                    xaxis_title='Pirates',
                    yaxis_title='Scores')
# you can save plotly graphs as html files to preserve interactivity and save webpage
#graph.write_html('pirate_scores.html')
```

```python
#customing plotly graphs with dict
graph=go.Figure()
graph.add_trace(go.Scatter(x=x1,y=y1,mode='markers',
                marker=dict(size=15,color='hotpink',
                opacity=1,
                line=dict(width=2,color='indigo'))))
```
