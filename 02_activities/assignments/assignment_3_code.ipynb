```python
#first load packages
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

#URL: https://open.toronto.ca/dataset/ttc-lrt-delay-data/
```

```python
#load data from local... ttc data but saved locally for delays (CSV) 
data = pd.read_csv('TTC_LRT_Delays.csv')
data.head()
```

```python
#First visualization - Min Delay by Station
min_delay_by_station = data.groupby('Station')['Min Delay'].min().sort_values() #sorting minimum delay values in order 
min_delay_by_station = min_delay_by_station[min_delay_by_station > 0] # I needed to filter only ones greater than 0
fig, ax = plt.subplots(figsize=(10, 6))
min_delay_by_station.plot(kind='barh', ax=ax, color='skyblue', edgecolor='black') # I am using pandas :)
ax.set_xlabel('Min Delay (minutes)')
ax.set_ylabel('Station')
ax.set_title('Minimum Delay by Station')
plt.tight_layout()
plt.savefig('min_delay_by_station.png', dpi=300, bbox_inches='tight')
```
